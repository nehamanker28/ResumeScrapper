```python
import pymupdf

# Open the PDF document
doc = pymupdf.open("/Users/nehamanker/Documents/GitHub/ResumeScrapper/data/resume_analysis/Neha_Manker_Resume.pdf")

# Iterate through pages and extract text
for page_num in range(len(doc)):
    page = doc.load_page(page_num)
    text = page.get_text()
    print(f"--- Page {page_num + 1} ---")
    print(text)
```
